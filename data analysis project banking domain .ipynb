```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## univariate analysis

```python
# 1. Define the list
my_cols = [
    "brid", "genderid", "iaid", "amountofcreditcards", 
    'nationality', 'occupation', 'feestructure', 
    'loyaltyclassification', 'propertiesowned', 'income band'
]

# 2. Run the loop
for i, col in enumerate(df[my_cols].columns):
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x=col)
    plt.title(f"Plot {i+1}: {col}")
    plt.show()
```

## Bivariate analysis

```python
# 1. Define the list
my_cols = [
    "brid", "genderid", "iaid", "amountofcreditcards", 
    'nationality', 'occupation', 'feestructure', 
    'loyaltyclassification', 'propertiesowned', 'income band'
]

# 2. Run the loop
for i, col in enumerate(df[my_cols].columns):
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x=col,hue="nationality")
    plt.title(f"Plot {i+1}: {col}")
    plt.show()
```

```python
# histplot of value count for different occupation 

for col in categorical_cols:
    if col=="occupation":
        continue
    plt.figure(figsize=(8,4))
    sns.histplot(df[col])
    plt.title("Histogram of occupation count")
    plt.xlabel(col)
    plt.ylabel("count")
    plt.show()
```

## Numerical Analysis

```python
 numerical_cols=["estimatedincome","superannuationsavings","creditcardbalance","bankloans","bankdeposits","checkingaccounts","savingaccounts","foreigncurrencyaccount","businesslending"]
plt.figure(figsize=(15,10))
for i, cols in enumerate(numerical_cols):
    plt.subplot(5,3,i+1)
    sns.histplot(df[cols],kde=True)
    plt.title(cols)
plt.show()
    
    
    
```

## HeatMaps

```python
 numerical_cols=["estimatedincome","superannuationsavings","creditcardbalance","bankloans","bankdeposits","checkingaccounts","savingaccounts","foreigncurrencyaccount","businesslending"]

correlation_matrix=df[numerical_cols].corr()
plt.figure(figsize=(12,12))
sns.heatmap(correlation_matrix,annot=True,cmap="crest",fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```
